# Convert heic to jpeg

```python
import os
import subprocess
```

```python

def convert_heic_to_jpg(directory):
    # Path to the ImageMagick executable
    magick_path = r'"C:\Program Files\ImageMagick-7.1.1-Q16-HDRI\magick"'

    # Loop through all files in the specified directory
    for filename in os.listdir(directory):
        if filename.lower().endswith(".heic"):
            # Construct full file path
            full_path = os.path.join(directory, filename)
            # Create a new filename for the jpeg
            new_filename = filename[:-5] + ".jpg"
            new_full_path = os.path.join(directory, new_filename)
            
            # Command to convert the file
            command = f'{magick_path} "{full_path}" "{new_full_path}"'
            
            # Run the command
            subprocess.run(command, shell=True)
            print(f'Converted {filename} to {new_filename}')

```

```python
import os
import subprocess

def convert_heic_to_jpg_delete_old_listed_recursive(file_full_path_list):
    # Path to the ImageMagick executable
    magick_path = r'"C:\Program Files\ImageMagick-7.1.1-Q16-HDRI\magick"'

    for file_full_path in file_full_path_list:
        ## print(file_full_path)
        if file_full_path.lower().endswith(".heic"):
            ## print("Ends with .heic")
            # Create a new filename for the jpeg
            new_full_path = file_full_path[:-5] + ".jpg"
            
            # Command to convert the file
            command = f'{magick_path} "{file_full_path}" "{new_full_path}"'
            
            # Run the command
            subprocess.run(command, shell=True)
            print(f'Converted {file_full_path} to {new_full_path}')

            # Delete the original .heic file
            os.remove(file_full_path)
            # print(f'Deleted original file {file_full_path}')

def find_all_files(directory):

    # Initial file list
    file_full_path_list = []

    for path in os.listdir(directory):
        # Construct full file path
        full_path = os.path.join(directory, path)
        # print(full_path)

        if os.path.isdir(full_path):
            # Loop through the next directory
            list_to_add = find_all_files(full_path)
            file_full_path_list = file_full_path_list +list_to_add
        elif os.path.isfile(full_path):
            # Add file to the list
            file_full_path_list = file_full_path_list + [full_path]
        
    return file_full_path_list

def add_heic_to_jpg_and_delete(directory):
    # Complete file list
    file_full_path_list = find_all_files(directory)

    convert_heic_to_jpg_delete_old_listed_recursive(file_full_path_list)
    
```

# JPG to PDF

```python
import os
from PIL import Image

def convert_jpgs_to_individual_pdfs(input_folder):
    # Create the output folder with "_pdf" suffix
    base_folder = os.path.abspath(input_folder)
    output_folder = base_folder + "_pdf"
    os.makedirs(output_folder, exist_ok=True)

    # Filter jpg/jpeg files
    jpg_files = [f for f in os.listdir(base_folder) if f.lower().endswith(('.jpg', '.jpeg'))]

    if not jpg_files:
        print("No JPG files found.")
        return

    for file_name in jpg_files:
        image_path = os.path.join(base_folder, file_name)
        image = Image.open(image_path).convert("RGB")

        # Remove extension and save PDF
        base_name = os.path.splitext(file_name)[0]
        output_pdf_path = os.path.join(output_folder, f"{base_name}.pdf")
        image.save(output_pdf_path, "PDF")

        print(f"Saved: {output_pdf_path}")
```

```python
import os
from PIL import Image

def convert_and_rotate_jpgs_to_pdfs(input_folder):
    base_folder = os.path.abspath(input_folder)
    output_folder = base_folder + "_pdf"
    os.makedirs(output_folder, exist_ok=True)

    jpg_files = [f for f in os.listdir(base_folder) if f.lower().endswith(('.jpg', '.jpeg'))]

    if not jpg_files:
        print("No JPG files found.")
        return

    for file_name in jpg_files:
        image_path = os.path.join(base_folder, file_name)
        image = Image.open(image_path).convert("RGB")

        # Rotate image 90 degrees clockwise
        rotated_image = image.rotate(-90, expand=True)  # -90 for clockwise

        # Save to PDF with same base name
        base_name = os.path.splitext(file_name)[0]
        output_pdf_path = os.path.join(output_folder, f"{base_name}.pdf")
        rotated_image.save(output_pdf_path, "PDF")

        print(f"Saved rotated PDF: {output_pdf_path}")
```

```python
import os
from PIL import Image
import pillow_heif

def convert_heic_folder_to_pdfs(input_folder):
    """
    Converts all .heic images in a folder to .pdf files
    and saves them into a new folder with '_pdf' suffix.
    """
    # Register HEIC support
    pillow_heif.register_heif_opener()

    # Get absolute path and folder name
    input_folder = os.path.abspath(input_folder)
    base_folder_name = os.path.basename(input_folder)
    output_folder = os.path.join(os.path.dirname(input_folder), base_folder_name + "_pdf")

    # Create output folder if it doesn't exist
    os.makedirs(output_folder, exist_ok=True)

    # Process all .heic files
    for filename in os.listdir(input_folder):
        if filename.lower().endswith(".heic"):
            heic_path = os.path.join(input_folder, filename)
            pdf_filename = os.path.splitext(filename)[0] + ".pdf"
            pdf_path = os.path.join(output_folder, pdf_filename)

            try:
                image = Image.open(heic_path).convert("RGB")
                image.save(pdf_path, "PDF", resolution=100.0)
                print(f"✅ Converted: {filename} → {pdf_filename}")
            except Exception as e:
                print(f"❌ Failed to convert {filename}: {e}")

    print(f"\nAll PDFs saved to: {output_folder}")
```

```python
import os
from PyPDF2 import PdfMerger

def combine_pdfs_in_folder(folder_path, output_filename="Combined_pdfs.pdf"):
    """
    Combines all PDF files in a folder into a single PDF.

    :param folder_path: Path to folder containing PDF files.
    :param output_filename: Name of the final combined PDF file.
    :return: Path to the combined PDF file.
    """
    folder_path = os.path.abspath(folder_path)
    output_path = os.path.join(folder_path, output_filename)

    # Get list of PDF files, sorted alphabetically
    pdf_files = sorted(
        [f for f in os.listdir(folder_path) if f.lower().endswith(".pdf")]
    )

    if not pdf_files:
        print("No PDF files found in the folder.")
        return None

    merger = PdfMerger()

    for pdf_file in pdf_files:
        file_path = os.path.join(folder_path, pdf_file)
        try:
            merger.append(file_path)
            print(f"🧷 Added: {pdf_file}")
        except Exception as e:
            print(f"❌ Failed to add {pdf_file}: {e}")

    merger.write(output_path)
    merger.close()

    print(f"\n📄 Combined PDF saved as: {output_path}")
    return output_path
```

```python
import pytesseract
from PIL import Image
import os
from pdf2image import convert_from_path
```

```python
def pdf_to_text(pdf_path, output_txt_path):
    pages = convert_from_path(pdf_path)
    with open(output_txt_path, 'w', encoding='utf-8') as f:
        for i, page in enumerate(pages):
            text = pytesseract.image_to_string(page)
            f.write(f"\n\n--- Page {i+1} ---\n{text}")

```
